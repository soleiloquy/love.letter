```python
import random

print("Welcome to my little guess-who game!")
fact_list = []
name_list = []
name = input("Who are you? Type your name: ")
name_list.append(name)
print(f"Hi, {name}!")
fact = input("What is an identifying fact about you, or a unique aspect of our relationship? ")
fact_list.append(fact)
num = random.randint(0, len(fact_list) - 1)
print(f"Thanks for your contribution! Now, guess who put this fact in the list: {fact_list[num]}")
guess = input("Guess here:") 
if guess == name_list.pop(): 
    print("You got it right!! Wow wow wow. Thanks for playing!") 
else: 
    print("You are not very right... Try again next time, and thanks for playing!")
```
