```python
def ask_who(): 
    name = input("Who are you? Name; ")
    return name 
reasons_list = {"Annabelle": "Hi Annabelle! Belly full of kisses yayayay", "Jennifer": "Hi little chud chud chud", "Howard": "Why are you here, you dense mf?"}
fact_list = {"Annabelle": "x"}


def one_fun_fact(reasons_list, name):
    fact = input("What's a fun fact about you? ") 
    fact_list[name] = fact

name = ask_who()
one_fun_fact(reasons_list, name)
print(reasons_list[name])
print(f"Here is our fact list: {fact_list}")
```
